```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
data = pd.read_csv("orders (1).csv")
data
```

```python
data = data[(data["Product"]=="MIS") & (data["Status"]=="COMPLETE")]
data
```

```python
data["Qty."] = data["Qty."].str.split("/").str[0].astype(int)
data
```

```python
data.info()
```

```python
data["Turnover"]= data["Avg. price"]*data["Qty."].round(2)
data
```

```python
data["Brokerage"] = (data["Turnover"]*0.03/100).round(2)
data
```

```python
data["Brokerage"] = data["Brokerage"].clip(upper=20)
data
```

```python
data["STT/CTT"] = (data["Turnover"] * 0.025 / 100 * (data["Type"] == "SELL")).round(2)
data
```

```python
data["ETC"] = (data["Turnover"]*0.00307/100).round(2)
data
```

```python
data["SEBI"] = (data["Turnover"]*10/10000000).round(2)
data
```

```python
data["Stamp charges"] = (data["Turnover"] * 0.003/100 * (data["Type"] == "BUY")).round(2)
data
```

```python
data["GST"] = ((data["Brokerage"] + data["SEBI"] + data["ETC"]) *18/100).round(2)
data
```

```python
data["Total charges"] = ((data["GST"] + data["Brokerage"] + data["SEBI"] + data["ETC"] + data["STT/CTT"] + data["Stamp charges"])).round(2)
data
```

```python
result1 = data
result1
```

```python
df1 = data.groupby(["Instrument","Type"],as_index=False)[["Qty."]].sum()
df1
```

```python
df2 = data.groupby(["Instrument","Type"],as_index=False)[["Avg. price"]].mean()
df2
```

```python
result2 = data.groupby(["Instrument","Type"],as_index=False).agg(
    {"Qty.":"sum",
     "Avg. price":"mean",
     "Turnover":"sum",
     "Brokerage":"sum",
     "STT/CTT":"sum",
     "ETC":"sum",
     "SEBI":"sum",
     "Stamp charges":"sum",
     "GST":"sum",
     "Total charges":"sum"})
result2
```

```python
result = result2.groupby(["Instrument","Type"])["Turnover"].sum().unstack()
result["Gross Pnl"] = result["SELL"] - result["BUY"]
result["Gross Pnl"]
```

```python
Total_charges = result2.groupby("Instrument")["Total charges"].sum()
Total_charges
```

```python
net_PnL = result["Gross Pnl"] - Total_charges
net_PnL
```

```python
Charges_on_Gross_PnL = (Total_charges / result["Gross Pnl"]) *100
Charges_on_Gross_PnL
```

```python
result3 = pd.DataFrame([result["Gross Pnl"],Total_charges,net_PnL,Charges_on_Gross_PnL])
result3
```

```python
result3.rename(index={"Unnamed 0":"net_PnL","Unnamed 1":"%Charges_on_Gross_PnL"},inplace=True)
result3
```

```python
result3.T
```

```python
with pd.ExcelWriter("kite Pnl.xlsx") as writer:
    result1.to_excel(writer,sheet_name="charges for Individual trade",index= False)
    result2.to_excel(writer,sheet_name="Stock Analysis",index= False)
    result3.to_excel(writer,sheet_name="Overall Summary",index= False)
```
